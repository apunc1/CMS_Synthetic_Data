Inventory local source files first. The preview helper reads only when explicitly called.

```python
from pathlib import Path

ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / 'src').is_dir())
RAW_DIR = ROOT / 'data' / 'raw'
files = sorted(path for path in RAW_DIR.iterdir() if path.is_file()) if RAW_DIR.exists() else []
[(path.name, round(path.stat().st_size / (1024 * 1024), 2)) for path in files]
```

```python
from typing import Union

import pandas as pd


def preview_csv(path: Union[str, Path], rows: int = 5) -> pd.DataFrame:
    """Preview a pipe-delimited CMS CSV without loading the full file."""
    return pd.read_csv(path, sep="|", nrows=rows, dtype=str)
```
