# Anomaly Model

Validate feature meaning, missingness, and scale before fitting an unsupervised baseline.

```python
from typing import Union

import pandas as pd
from sklearn.ensemble import IsolationForest


def fit_anomaly_model(
    features: pd.DataFrame, contamination: Union[str, float] = "auto"
) -> tuple[IsolationForest, pd.Series]:
    numeric = features.select_dtypes(include="number").dropna()
    if numeric.empty:
        raise ValueError("Provide at least one complete numeric feature row.")
    model = IsolationForest(contamination=contamination, random_state=42)
    model.fit(numeric)
    labels = pd.Series(model.predict(numeric), index=numeric.index, name="anomaly_label")
    return model, labels
```
